```python
import numpy as np
import pandas as pd

#Defining path to file
path = "/mnt/c/Users/josem/OneDrive/Documents/Dummy_data/new_dummy_data.csv" # r'/path' for windows

# using pandas pd.read_csv(path) I will load in the data
df = pd.read_csv(path)
# Choose three wells to be hot.
hot_wells = [10, 1, 5]

# Start every well at a low background concentration.
df["PCB"] = 5.0

# Give the three selected wells high concentrations.
df.loc[hot_wells, "PCB"] = [1200.0, 1050.0, 850.0]

df
```

```python
rng = np.random.default_rng(42)

cold_wells = ~df.index.isin(hot_wells)
df.loc[cold_wells, "PCB"] = rng.uniform(
    low=0.01,
    high=10.0,
    size=cold_wells.sum(),
)

df
```

```python
df["PCB"] = df["PCB"].astype(int)
df
```

```python
df.to_csv('/mnt/c/Users/josem/OneDrive/Documents/Dummy_data/new_dummy_data2.csv', index=False)
```
